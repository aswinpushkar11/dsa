#### _Delete from Tail :Pop_ ###

```python
# Creating a Node Class
class Node:

    def __init__(self,value):
        self.data = value 
        self.next = None


# Creating a LinkedList
class LinkedList:
    
    def __init__(self):
        #Empty linked list
        self.head = None
        #No of nodes in the linked list
        self.n = 0     


    #magic method for the length of the linkedlist
    def __len__(self):
        return self.n   


    # Inserting a new value in the head
    def insert_head(self,value):
        #new node
        new_node =   Node(value)

        #create connection
        new_node.next = self.head

        #reassign head
        self.head =new_node

        #increment n
        self.n +=1  


    #printing the LinkedList 1-> 2-> 3 ....
    def __str__(self):
        curr = self.head #address of the head object
        result = ''
        while curr != None:
            result = result +str(curr.data)+'->'
            curr = curr.next
        return result[:-2]
    
    #Append
    def append(self,value):
        new_node = Node(value)
        curr =self.head
        print(curr)
        while curr.next != None:
            curr = curr.next
        #You are at the last node
        curr.next = new_node

    def insert_after(self,after,value):
        new_node = Node(value)
        curr = self.head

        while curr != None :
            if curr.data == after:
                break
            curr = curr.next


        #Case1 break -curr is not None

        if curr != None:
            new_node.next = curr.next
            curr.next = new_node
            self.n = self.n+1
        #Case2 curr is None
        else:
            return 'Item not found'
        

    def clear(self):
        self.head = None
        self.n = 0

    def delete_head(self):
        if self.head == None:
            return "Empty LinkedList"
        self.head = self.head.next
        self.n-=1

    def pop(self):
        if self.head == None:
            return 'Empty LinkedList'
        curr = self.head
        #check the count of items in LL
        if curr.next == None:
            return self.delete_head()
            
        while curr.next.next != None:
            curr = curr.next
        
        #curr is the second last node
        curr.next  = None
        self.n -= 1
        
```

```python
L=LinkedList()
L.insert_head(1)
L.insert_head(2)
L.insert_head(3)
L.insert_head(4)
print(L)
```
