### _Search Algorithm_ ###

##### _Linear Search Algorithm_ #####

```python
# - Like a brute force approach

def linear_search(arr,item):
    for i in range(len(arr)):
        if arr[i] == item:
            return f'The number {item} is at Index {i}'
    return -1
    
```

```python
arr = [23,45,56,45,24,667,78]
linear_search(arr,56)
```

##### _Binary Search Algorithm_ #####


```python
#Applicable for sorted arrays

def binary_search(arr, low, high, item):
    print('low',low,'high',high,end=' ')
    if low <= high:
        #search
        mid = (low+high)//2
        print('mid value',mid)
        
        if arr[mid] == item:
            return f'The index of the number {item} is {mid}'
        elif arr[mid] > item:
            return binary_search(arr,low,mid-1,item)
        else:
            return binary_search(arr,mid+1,high,item)
    
    else:
        return -1
```

arr = [11,23,34,45,56,67,78,89,100]
print(binary_search(arr,0,len(arr)-1,100))

```python
# is_sorted

def is_sorted(arr):
    sorted = True
    for i in range(len(arr)-1):
        if arr[i]> arr[i+1]:
            sorted = False
    return sorted
```

```python
arr=[1,2,8,4,5]
is_sorted(arr)
```

```python
#monkey sorting
import time ,random
def monkey_sort(arr):
    
    while not is_sorted(arr):
        time.sleep(1)
        random.shuffle(arr)
        print(arr)
    print('The sorted array :',arr)

arr = [1,5,4,3]
monkey_sort(arr)
```

#### _Bubble Sort_ ####
- Stable(Preserves the order of the similar element)
- Flag addition makes it adaptive (If the array is already sorted need not perform (n-1) passes)

```python
arr = [14,99,34,45,23,91,13]
arr
```

```python
def bubble_sort(arr):
    #no of passes
    for i in range (len(arr)-1):
        #comparison
        for j in range(len(arr)-1-i):
            #swapping
            if arr[j]> arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]

    return arr
```

```python
print(bubble_sort(arr))
```

##### _Adaptive Bubble Sort_ #####

```python
sorted_arr = [1,2,3,4,5]
```

```python
def bubble_sort(arr):
    #no of passes
    for i in range (len(arr)-1):
        #flag added for the sorted check in first pass
        flag=0
        #comparison
        for j in range(len(arr)-1-i):
            #swapping
            if arr[j]> arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
                flag=1
        print('Flag Value:',flag)
        if flag == 0:
            print('The Array is alread Sorted,Hence breaking in the first Pass')
            break

    return arr
```

```python
print(bubble_sort(sorted_arr))
```
