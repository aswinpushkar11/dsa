```python
#creating python list using c arrays
import ctypes

class MyList:

    def __init__(self):
        #max number of items that can be stored
        self.size = 1

        #number of items at present
        self.n = 0
        
        #create a c type array with size = self.size
        self.A = self.__make_array(self.size)

    #length of the array
    def __len__(self):
        return self.n

    def append(self,item):
        #check if size of the array and number of items present are same
        if self.n == self.size:
            #resize
            self.__resize(self.size*2)

        self.A[self.n] = item
        self.n = self.n + 1

    def __resize(self,new_capacity):
        #create a new array with new capacity
        B= self.__make_array(new_capacity)
        self.size = new_capacity
        #copy the contents of A to B'
        for i in range(self.n):
            B[i] = self.A[i]
        #reassign A
        self.A = B


    #function for creating arrays
    def __make_array(self,capacity):
        #create a c type array(static,referential) with size capacity
        return (capacity*ctypes.py_object)()
```

```python
#creating python list using c arrays
import ctypes

class MyList:

    def __init__(self):
        #max number of items that can be stored
        self.size = 1

        #number of items at present
        self.n = 0
        
        #create a c type array with size = self.size
        self.A = self.__make_array(self.size)

    #Printing the list
    def __str__(self):
        result = ''
        for i in range(self.n):
            result = result +str(self.A[i])+','
        # returning the string with the bracket and removing the last comma
        return '[' + result[:-1] + ']'

    #length of the array
    def __len__(self):
        return self.n
    
    #magic method for creating Index
    def __getitem__(self,index):
        if 0 <= index <= self.n:
            return self.A[index]
        else:
            return "IndexError - Index out of Range"

    def append(self,item):
        #check if size of the array and number of items present are same
        if self.n == self.size:
            #resize
            self.__resize(self.size*2)

        self.A[self.n] = item
        self.n = self.n + 1

    def insert(self,pos,item):
        #check if you have an extra space for new item
        if self.n == self.size:
            self.__resize(self.size*2)

        for i in range(self.n,pos,-1):
            self.A[i] = self.A[i-1]
        
        self.A[pos] = item
        self.n = self.n+1

    #remove uses finding position of item followed by delete logic
    def remove(self,item):
        pos = self.find(item)
        if type(pos) == int:
            self.__delitem__(pos)
        else:
            return pos


    #pop function
    def pop(self):
        if self.n == 0:
            return 'Empty string'
        
        print(self.A[self.n-1])
        self.n = self.n - 1

    #clear function
    def clear(self):
        self.n = 0
        self.size = 1

    
    #delete
    def __delitem__(self,pos):
        if 0 <= pos <= self.n:
            for i in range(pos,self.n-1):
                self.A[i] = self.A[i+1]
            self.n =self.n-1

  
    def __resize(self,new_capacity):
        #create a new array with new capacity
        B= self.__make_array(new_capacity)
        self.size = new_capacity
        #copy the contents of A to B'
        for i in range(self.n):
            B[i] = self.A[i]
        #reassign A
        self.A = B

    def find(self,item):
        for i in range(self.n):
            if self.A[i] == item:
                return i
        return 'ValueError-not in List'


    #function for creating arrays
    def __make_array(self,capacity):
        #create a c type array(static,referential) with size capacity
        return (capacity*ctypes.py_object)()
```
